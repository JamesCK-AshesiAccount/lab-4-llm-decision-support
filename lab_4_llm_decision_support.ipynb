```python
# @title
import os
from google.colab import userdata

API_KEY = userdata.get("GROQ_API_KEY")

if not API_KEY:
    raise ValueError("GROQ_API_KEY was not found. Check your Colab Secrets.")

print("API key loaded successfully.")
```

```python
#making the LLM client
from openai import OpenAI

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

MODEL = "llama-3.3-70b-versatile"

print("Client ready.")
```

```python
#helper function
def ask_llm(
    user_prompt,
    system_prompt="You are a helpful assistant.",
    temperature=0.7,
    max_tokens=500
):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )

    return response
```

```python
#make the first API call
response = ask_llm(
    "What is artificial intelligence? Explain it in two sentences."
)

print("Response:")
print(response.choices[0].message.content)

print("\nToken usage:")
print(response.usage)
```

1. What is the difference between the system and user roles? Give an example of something that belongs in each.

The system role defines the overall behavior, role, and constraints that the LLM should follow. For example, a system message could instruct the model to act as a neutral assistant to a microfinance loan officer and not invent information. The user role contains the specific request or information that the model needs to respond to, such as asking it to summarize a particular loan application.

2. What is a token, roughly? Why do API providers bill per token rather than per request?

A token is a small unit of text processed by an LLM. It can represent a whole word, part of a word, punctuation, or another piece of text. API providers generally bill per token because the computational resources required depend on the amount of text the model processes and generates, so token usage provides a better measure of the actual work performed than simply counting requests.

```python
#defining the test question
question = "Suggest a name for a savings product for market traders in Accra."
```

```python
print("===== TEMPERATURE 0.0 =====\n")

temperature_0_results = []

for i in range(5):
    response = ask_llm(
        question,
        temperature=0.0,
        max_tokens=100
    )

    answer = response.choices[0].message.content
    temperature_0_results.append(answer)

    print(f"Run {i+1}: {answer}")
    print()
```

```python
print("===== TEMPERATURE 1.2 =====\n")

temperature_12_results = []

for i in range(5):
    response = ask_llm(
        question,
        temperature=1.2,
        max_tokens=100
    )

    answer = response.choices[0].message.content
    temperature_12_results.append(answer)

    print(f"Run {i+1}: {answer}")
    print()
```

What did you observe at each temperature? For the loan decision-support system you are about to build, which temperature regime is appropriate, and why?

At temperature 0.0, the responses were more consistent and tended to produce similar suggestions across the different runs. At temperature 1.2, the responses showed more variation in the names and wording produced by the model. This demonstrates how temperature can be used to control the randomness of an LLM’s responses.

For the loan decision-support system, a low temperature, particularly 0.0, is more appropriate for tasks such as structured information extraction because the system needs to be consistent and reproducible. A higher temperature would introduce more variation, which could make the extraction of factual information less reliable. Lower temperature is therefore preferable when accuracy and consistency are more important than creativity.

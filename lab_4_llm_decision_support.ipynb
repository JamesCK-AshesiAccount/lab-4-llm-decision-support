````python
#making the LLM client
from openai import OpenAI

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

MODEL = "llama-3.3-70b-versatile"

print("Client ready.")
````

````python
#helper function
def ask_llm(
    user_prompt,
    system_prompt="You are a helpful assistant.",
    temperature=0.7,
    max_tokens=500
):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )

    return response
````

````python
#make the first API call
response = ask_llm(
    "What is artificial intelligence? Explain it in two sentences."
)

print("Response:")
print(response.choices[0].message.content)

print("\nToken usage:")
print(response.usage)
````

1. What is the difference between the system and user roles? Give an example of something that belongs in each.

The system role defines the overall behavior, role, and constraints that the LLM should follow. For example, a system message could instruct the model to act as a neutral assistant to a microfinance loan officer and not invent information. The user role contains the specific request or information that the model needs to respond to, such as asking it to summarize a particular loan application.

2. What is a token, roughly? Why do API providers bill per token rather than per request?

A token is a small unit of text processed by an LLM. It can represent a whole word, part of a word, punctuation, or another piece of text. API providers generally bill per token because the computational resources required depend on the amount of text the model processes and generates, so token usage provides a better measure of the actual work performed than simply counting requests.

Lower temperature is therefore preferable when accuracy and consistency are more important than creativity.

````python
import json
import pandas as pd
````

````python
EXTRACT_SYSTEM = """
You are a data extraction assistant for a microfinance institution.

Extract factual information from loan application letters.

Return ONLY a valid JSON object.
Do not include markdown, explanations, comments, or code fences.

The JSON object must contain EXACTLY these keys:

{
  "applicant_name": "string",
  "amount_ghs": "number",
  "purpose": "string",
  "monthly_profit_ghs": "number or null",
  "has_collateral_or_guarantor": "boolean",
  "repayment_months": "number or null"
}

Rules:
- Use only information explicitly stated in the letter.
- If a field is not stated, use null.
- Do not guess or infer missing information.
- amount_ghs must be the requested loan amount.
- monthly_profit_ghs must be the stated monthly profit, or null if it is not stated.
- has_collateral_or_guarantor is true if the letter states that collateral or a guarantor exists.
- repayment_months must be the stated repayment period in months, or null if it is not stated.
- Preserve the meaning of the stated loan purpose.
"""
````

````python
EXAMPLE_LETTER = """
My name is Ama Owusu and I operate a small bakery in Cape Coast.
I am requesting GHS 6,000 to purchase a commercial oven.
My bakery earns a monthly profit of GHS 1,200.
I can repay the loan over 10 months.
My brother will act as my guarantor.
"""

EXAMPLE_JSON = {
    "applicant_name": "Ama Owusu",
    "amount_ghs": 6000,
    "purpose": "purchase a commercial oven",
    "monthly_profit_ghs": 1200,
    "has_collateral_or_guarantor": True,
    "repayment_months": 10
}
````

````python
#Build the final extraction prompt
EXTRACT_PROMPT = """
Here is an example of the required extraction format.

APPLICATION:
{example_letter}

EXPECTED JSON:
{example_json}

Now extract information from the following loan application.

Return ONLY the JSON object.

APPLICATION:
{letter_text}
"""
````

````python
def extract_fields(letter_text):
    try:
        prompt = EXTRACT_PROMPT.format(
            example_letter=EXAMPLE_LETTER,
            example_json=json.dumps(EXAMPLE_JSON, indent=2),
            letter_text=letter_text
        )

        response = ask_llm(
            prompt,
            system_prompt=EXTRACT_SYSTEM,
            temperature=0.0,
            max_tokens=300
        )

        raw_output = response.choices[0].message.content.strip()

        # Remove markdown JSON fences if present
        if raw_output.startswith("```json"):
            raw_output = raw_output[7:]

        elif raw_output.startswith("```"):
            raw_output = raw_output[3:]

        if raw_output.endswith("```"):
            raw_output = raw_output[:-3]

        raw_output = raw_output.strip()

        result = json.loads(raw_output)

        return result

    except json.JSONDecodeError:
        print("Warning: Model returned invalid JSON.")
        print("Raw output:", raw_output)
        return None

    except Exception as e:
        print(f"Warning: Extraction failed: {e}")
        return None
````

````python
result = extract_fields(LETTERS["L001"])

print(json.dumps(result, indent=2))
````

````python
extracted_results = {}

for letter_id, letter_text in LETTERS.items():
    print(f"Extracting {letter_id}...")
    extracted_results[letter_id] = extract_fields(letter_text)

print("\nExtraction complete.")
````

````python
BRIEF_SYSTEM = """
You are an assistant to a microfinance loan officer.

Your role is to provide decision-support information, not to make the final
loan decision.

Analyze the loan application using only information contained in the
application and the extracted structured data.

Your response must contain exactly four sections:

1. Strengths
- List positive factors explicitly supported by the application.

2. Risks / Red Flags
- List financial, business, repayment, collateral, or other relevant risks.
- Do not invent risks that are not supported by the information provided.

3. Missing Information
- Identify important information the loan officer should request before
  making a decision.

4. Suggested Next Step
- Recommend an appropriate human-review action such as requesting documents,
  inviting the applicant for an interview, or flagging the application for
  senior review.
- Do NOT say "approve" or "reject".
- The final loan decision must always be made by a human loan officer.

Be factual, neutral, and concise.
"""
````

````python
BRIEF_PROMPT = """
Prepare a decision-support brief for the following loan application.

LOAN APPLICATION:
{letter_text}

EXTRACTED STRUCTURED DATA:
{extracted_data}

Provide:

1. Strengths
2. Risks / Red Flags
3. Missing Information
4. Suggested Next Step

Remember that the final decision must be made by a human loan officer.
Do not output "approve" or "reject".
"""
````

````python
def generate_brief(letter_text, extracted_data):
    response = ask_llm(
        BRIEF_PROMPT.format(
            letter_text=letter_text,
            extracted_data=json.dumps(extracted_data, indent=2)
        ),
        system_prompt=BRIEF_SYSTEM,
        temperature=0.0,
        max_tokens=500
    )

    return response.choices[0].message.content
````

Prompting as engineering

Prompt engineering is similar to tuning model hyperparameters because both involve testing different configurations and evaluating their effects on performance. In Lab 3, I changed model parameters such as learning rate and network settings to improve the model. In this lab, I changed instructions, constraints, examples, and temperature to improve the LLM’s outputs. The main difference is that prompt engineering changes how a pre-trained foundation model is instructed rather than changing the model’s learned parameters.

Trust

I would not trust this system to run completely unattended for loan decisions. For this task, calling an API is preferable to training a model from scratch because a foundation model has already been trained on a very large amount of data and can understand and generate natural language without requiring a large labelled dataset or significant training resources. Building a specialized model would require considerably more data, computing resources, development time, and maintenance. However, training or fine-tuning a model could make sense when an organization has a large amount of domain-specific data, requires specialized behavior, needs greater control over the model, or has privacy and deployment requirements that make sending data to an external API inappropriate.
